### Outliers Analysis

```python
def outliers_handling(column_name):
    q1=np.percentile(df[column_name],25)
    q2=np.percentile(df[column_name],50)
    q3=np.percentile(df[column_name],75)
    IQR=q3-q1
    lb=q1-1.5*IQR
    ub=q3+1.5*IQR
    # ======================================================================================
    cond= (df[column_name]<lb) | (df[column_name]>ub)
    true=df[column_name].median()
    false=df[column_name]
    df[column_name]=np.where(cond,true,false)
    return df[column_name]
def outliers_len(column_name):
    q1=np.percentile(df[column_name],25)
    q2=np.percentile(df[column_name],50)
    q3=np.percentile(df[column_name],75)
    IQR=q3-q1
    lb=q1-1.5*IQR
    ub=q3+1.5*IQR
    cond= (df[column_name]<lb) | (df[column_name]>ub)
    outliers_df=df[[column_name]][cond]
    # len(outliers_df)
    print(f"The length of the outliers is : {len(outliers_df)}")
```

```python
def box_plot(colum_name):
    plt.figure(figsize=(10,5))
    df[colum_name].plot(kind='box',vert=False)
    plt.title(f"{colum_name} Box plot")
    plt.show()
```

```python
LR=LogisticRegression()
LR.fit(X_train,y_train)
```

```python
param_grid = {
    "penalty": ['l2', 'none'],  # 'l1' and 'elasticnet' only work with 'saga'
    "dual": [False],  # dual=True only works with 'liblinear'
    "tol": [1e-4, 1e-3, 1e-2],
    "C": [0.01, 0.1, 1.0, 10.0, 100.0],
    "fit_intercept": [True, False],
    "intercept_scaling": [1, 2, 5],
    "class_weight": [None, 'balanced'],
    "random_state": [42],  # Optional: fix for reproducibility
    "solver": ['lbfgs', 'saga'],
    "max_iter": [100, 200, 500],
    "verbose": [0],
    "warm_start": [False],
    # Remove l1_ratio when not using elasticnet
    # "l1_ratio": [None],  # Remove for now
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
Grid_search=GridSearchCV(LR,
                        param_grid,
                        cv=5,
                        scoring='accuracy',
                        verbose=True)

Grid_search    
```

```python
Grid_search.fit(X_train,y_train)
```

```python
best_LR_Model=Grid_search.best_estimator_
best_LR_Model
```

```python
y_pred_tune=best_LR_Model.predict(X_test)
y_pred_tune
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
acc=round(accuracy_score(y_test,y_pred_tune),2)
prec=round(precision_score(y_test,y_pred_tune),2)
rec=round(recall_score(y_test,y_pred_tune),2)
f1=round(f1_score(y_test,y_pred_tune),2)
Grid_LR_Mtrics=[acc,prec,rec,f1]
pd.DataFrame(Grid_LR_Mtrics,columns=['Grid_LR_Mtrics'],index=["accuracy_score","precision_score","recall_score","f1_score"])
```
